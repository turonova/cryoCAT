# Symmetry groups

* General setting: finite subgroups of the group of rotations $SO(3)$.
* Up to conjugation, all finite subgroups of $SO(3)$ are:

    - **Cyclic symmetry groups $C_n$**: Rotational symmetries of regular $n$-gons (regular polygons with $n$ vertices)
    - **Dihedral groups $D_n$**: Symmetries of regular $n$-gons
    - **Tetrahedral group $T\cong A_4$**: Rotational symmetries of the tetrahedron
    - **Octahedral group $O\cong S_4$**: Rotational symmetries of the cube / octahedron
    - **Icosahedral group $I\cong A_5$**: Rotational symmetries of the icosahedron / dodecahedron

* All of these groups are handled within cryoCAT.

* The class SymmGroup is part of the `cryocat/utils/symmetry.py` module. 
    - It handles operations common to all groups that are being considered ($C_n, D_n, T, O, I$).
    - Every group has generators; these are rotations the iterated products of which yield all respective symmetries.

## Initialize groups
In the following section, examples of how to handle instances from this class are shown. <br>
NOTE: For all the functions displayed, it is assumed that the `cryocat.utils.symmetry` module is imported:

```python
from cryocat.utils.symmetry import *
```

For any of these groups, an object has to be initialized. The resulting instance is corresponds to a default embedding of the group into the group of rotation matrices $SO(3)$. <br>
For $C_n$ and $D_n$, these default options correspond to polygons in the $xy$-plane with vertices given by $(\cos(\frac{2\pi k}{n}), \sin(\frac{2\pi k}{n}), 0)$ for $k=0,...,n-1$. <br>
All symmetry groups are listed in the `SYMMETRY_GROUPS` dictionary:

```python
SYMMETRY_GROUPS
```

An instance of the icosahedral group can thus be obtained by `I = IcosahedralGroup()` or as done below.

```python
I = SYMMETRY_GROUPS['I']

print(f"The icosahedral group consists of {I.order} rotations.")
```

## Compute non-standard embeddings
As mentioned earlier, the above commmand produces a default embedding of the initialized group. <br>
For a custom embedding, i.e. symmetries that fix a different icosahedron (or, in the case of $C_n$-symmetry, say, rotations around an axis different from the $z$-axis), <br>
the group generators are conjugated by a rotation to alter the embedding. <br>

For a generating rotation matrix $R\in SO(3)$, this means that its rotation is composed with another rotation $C$ as below: <br>
$R_{\text{conjugated}}=CRC^T$ <br>

Such matrices $C$ can be designed to align different symmetry axes. <br>
In the example below, the cube that is acted on by the symmetry group $O$ is rotated in such a way that the axis $\frac{1}{\sqrt{2}}(1,1,0)^T$ is aligned with the $x$-axis. <br>
For this purpose, a rotation $C$ can be computed with a help function that computes a rotation based on an input axis and an output axis. 

```python
import numpy as np

input_axis = np.array([1,1,0])
# Axis is normalized automatically!

output_axis = np.array([1,0,0])

conjugation_matrix = compute_conjugation_matrix(in_axis = input_axis, out_axis = output_axis)

angles_df = get_symmetry_angles(
    symmetry = 'O',
    euler_convention = "zxz",
    degrees = True,
    conjugation_matrix = conjugation_matrix,
    return_df = True,
    out_path = None
)
```

For this examle, the `out_path`-variable was set to `None`, but the underlying data frame can be saved as a `csv` file. <br>
For this purpose, it is merely displayed:

```python
display(angles_df)
```
